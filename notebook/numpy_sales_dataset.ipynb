```python
#Find the average quantity of records where sales are greater than 3000
#Find the highest quantity among those records
quantity=data[:,1].astype(int)
sales=data[:,4].astype(int)
average=np.mean(quantity[sales>3000])
print("Average Quantity:",average)
print("Highest quantity:",np.max(quantity[sales>3000]))
index=np.where((sales>3000)&(quantity==np.max(quantity[sales>3000])))[0][0]
print("Record of highest quantity:",data[index])
```
