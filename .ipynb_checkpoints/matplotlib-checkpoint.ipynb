```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#pyplot functions
#1. plot()---> used to create line plots by connecting data points
    #syntax-->plt.plot(x,y,color='color name',linestyle='line_style',linewidth=value,marker='marker symbol',label='label name')
#2.xlabel() and ylabel()
#3.title()-->headig at top of graph
#4.Legend()--> shows what each line and marker is for in rectangular box in top
          #loc-upperleft( top left corner) and lower right(bottom right corner)
#5.grid()-makes grid in background and also can pass parameters inside it 
#6. xlim() and ylim()--->Axis limits.. used to control ranges of x axis and y axis
#7.xticks() and yticks()--->axis ticks-->give meaningful labeling to default numeric ticks
#8.plt.show()--->display the final plot designed to notebook cell or screen 
months=[1,2,3,4]
sales=[1000,1500,1200,1800]
plt.plot(months,sales,color='blue',linestyle='--',linewidth=2,marker='o',label='2025 sales data')
plt.xlabel('Months')
plt.ylabel('Sales per months')
plt.title('Monthly Sales Data Report')
plt.legend(loc="upper left")
plt.grid(color='gray',linestyle=':',linewidth=1)
plt.xlim(1,4)# limit the range or x axis
plt.ylim(0,2000)#limit the range of y axis
plt.xticks([1,2,3,4],['M1','M2','M3','M4'])
plt.show()
```

```python
#plotting Charts(bar chart,pie chart,histograms)
#bar chart-->Category comparison using bars/data analysis 
#pie chart--->Proportion illustration/whole representation
#Histograms--> Numerical Distribution/Data insights
```

```python
#Bar chart
#syntax--> plt.bar( x, height,color='color name',width=value, label='label name')
#barchart can be horizontal and vertical
#barh()--> for horizontal bar chart
product=['A' ,'B' ,'C','D']
sales=[1000,1500,800,1200]
plt.bar(product,sales,color='orange',label='Sales 2025')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Product Based Sales')
plt.legend()
plt.show()
```

```python
#pie chart--> proportion based illustration(slices of circle and shows how each part contribute to circle)
#syntax-->plt.pie(values,labels=label_list,colors=color_list,autopct='%1.1f%%')
regions=['North','South','East','West']
revenue=[3000,2000,1500,1000]

plt.pie(revenue,labels=regions,autopct='%1.1f%%',colors=['gold','skyblue','lightgreen','coral'])
plt.title('Reveneue Contribution By Region')
plt.show()
```

```python
#plottimg Histograms-->group continuous data and show frequency of each data in each range
#to show distribution of continuous data by dividing into ranges(bins)
#plt.hist(data,bins=number_of_bins,color='colorname',edgecolor='black')
#height of bar-->no.students that fall in that range

scores=[45,67,89,30,69,56,76,49,75,43,23,75,70,67,42,99,98,77,88,84]
plt.hist(scores,bins=5, color='green',edgecolor='black')
plt.xlabel('Score range')
plt.ylabel('Number of Students')
plt.title('Score distribution of Students')
plt.show()
```

```python
#Scatter Plot--> to show relationship between two numerical variables,displayed using dots on a grid
#Graph where each point represents one observation with two values(x and y)
#syntax-->plt.scatter(x,y,color='color_name',marker='marker style',label='label name')
#used to find 1.correlation between two variables 2.
hours_studies=[1,2,3,4,5,6,7,8]
exam_scores=[50,55,66,78,95,32,56,100]
plt.scatter(hours_studies,exam_scores,color='green',marker='o',label='Student Data')
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Relationship between study time and Exam Score")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#subplots---> create multiple plots in one figure
x=[1,2,3,4]
y=[10,20,15,25]
plt.subplot(1,2,1) # 1 row,2 column,1 position
plt.plot(x,y)
plt.title("Line Chart")


plt.subplot(1,2,2) #1 row,2 column,2 position
plt.bar(x,y)
plt.title('Bar Chart')
plt.tight_layout()
plt.show()
#easy to compare both plots in same window
```

```python
#fig,ax=plt.subplots(nrows,ncols,figsize(width,height) 
#object oriented way to create multiple plots in one window
fig,ax=plt.subplots(1,2,figsize=(10,5))
x=[1,2,3,4]
y=[10,20,15,25]

ax[0].plot(x,y,color="blue")
ax[0].set_title("Line plot")
ax[1].bar(x,y,color="orange")
ax[1].set_title("Bar Chart")
fig.suptitle('Comparison of line and bar charts')#all figures subtitle added
plt.tight_layout()
plt.show()
```

```python
#plt.tight_layout()---automatic adjustment according to window size
# use it before plt.show()
```

```python
#Saving Figures
#why we need to save figures-->save your graph figures to image file
#savefig()-->save your current figure into image file(jpg,png etc)

# plt.savefig("filename.extension',dpi=value,bbox_inches='tight')-->Syntax
#dpi->dots per inch(to save high quality images when set high)
#bbox_inches->crop white spaces in image automatically

months=[1,2,3,4]
sales=[1000,1500,1200,1800]
plt.plot(months,sales,color='blue',linestyle='--',linewidth=2,marker='o',label='2025 sales data')
plt.xlabel('Months')
plt.ylabel('Sales per months')
plt.title('Monthly Sales Data Report')
plt.savefig('line_plot.png',dpi=300,bbox_inches="tight")#use savefig before show function
plt.show()
```
